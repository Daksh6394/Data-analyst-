```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data= pd.read_csv('Barriers in Manufacturing Industry (Responses) - Form Responses 1.csv')
```

```python
data.head()
```

```python
data = data.drop(columns=['Environmental Sustainability Issues.1'])
```

```python
data.head()
```

```python
data = data.drop(columns=['If any other Barrier observed'])
```

```python
data = data.drop(columns=['Email'])
```

```python
data.head()
```

```python
data = data.drop(columns=['Email Address'])
```

```python
data.head()
```

```python
normalized_data = (data - data.min()) / (data.max() - data.min())
```

```python
print(normalized_data)
```

```python
def grey_relational_coefficient(x, y, rho=0.5):
    diff = np.abs(x - y)
    min_diff = np.min(diff)
    max_diff = np.max(diff)
    return (min_diff + rho * max_diff) / (diff + rho * max_diff)
```

```python
target = normalized_data.iloc[:, -1]
```

```python
grc_values = []
```

```python
for column in normalized_data.columns[:-1]:  # Excluding target variable column
    factor = normalized_data[column]
    grc = grey_relational_coefficient(factor, target)
    grc_values.append(np.mean(grc))  # Take the average GRC for each factor
```

```python
grc_df = pd.DataFrame({
    'Factor': normalized_data.columns[:-1],
    'Grey Relational Coefficient': grc_values
})
```

```python
grc_df_sorted = grc_df.sort_values(by='Grey Relational Coefficient', ascending=False)
```

```python
plt.figure(figsize=(10, 6))
plt.barh(grc_df_sorted['Factor'], grc_df_sorted['Grey Relational Coefficient'], color='skyblue')
plt.xlabel('Grey Relational Coefficient')
plt.ylabel('Factors')
plt.title('Grey Relational Coefficient for Supply Chain Resilience Factors')
plt.show()
```
